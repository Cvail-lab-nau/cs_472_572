## 0) Setup & Data Load

```python

# Setup (no clustering code here)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_context("notebook")

# 🔗 Data source
URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

# Load
df = pd.read_csv(URL)

print("Shape:", df.shape)
display(df.head())
display(df.describe(include='all').T)
```


## 1) Choose Features (edit the list below)
Pick **2–4 numeric features** to cluster on. Start with all four; later try a 2‑feature subset for easier plots.


```python

#Edit this list to your chosen features
chosen = ['sepal_length','sepal_width','petal_length','petal_width']

X = df[chosen].copy()
print("Chosen features:", chosen, "| shape:", X.shape)
display(X.head())
```


## 2) Preprocessing: minimal clean + standardize
We’ll drop any missing rows (Iris has none typically) and **standardize** the chosen features so each contributes fairly to distance.


```python

# Minimal cleaning: drop missing rows (robustness)
X_clean = X.dropna().copy()

# Standardize (mean=0, std=1)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_clean)

# DataFrame version for easy plotting
Xs = pd.DataFrame(X_scaled, columns=chosen, index=X_clean.index)

print("After cleaning:", X_clean.shape, "| Scaled array shape:", X_scaled.shape)
display(Xs.describe().T.round(3))
```


## 3) Visualization (provided)
Use these to understand the feature space before clustering.


```python

# Pairplot in RAW space
sns.pairplot(X, corner=True, diag_kind='hist')
plt.suptitle("Pairplot — raw features", y=1.02)
plt.show()
```

```python

# Pairplot in SCALED space
sns.pairplot(Xs, corner=True, diag_kind='hist')
plt.suptitle("Pairplot — standardized features", y=1.02)
plt.show()
```

```python

# Boxplots (raw vs scaled)
fig, ax = plt.subplots(1,2, figsize=(10,4))
sns.boxplot(data=X, ax=ax[0]);   ax[0].set_title("Raw features")
sns.boxplot(data=Xs, ax=ax[1]);  ax[1].set_title("Standardized features")
plt.tight_layout(); plt.show()
```

```python

# 2D scatter of first two chosen features (raw vs scaled)
f0 = chosen[0]
f1 = chosen[1] if len(chosen) > 1 else chosen[0]

plt.figure(figsize=(6,5))
plt.scatter(X[f0], X[f1], s=20, alpha=0.8)
plt.xlabel(f0); plt.ylabel(f1); plt.title("Raw space (first two features)")
plt.show()

plt.figure(figsize=(6,5))
plt.scatter(Xs[f0], Xs[f1], s=20, alpha=0.8)
plt.xlabel(f0); plt.ylabel(f1); plt.title("Standardized space (first two features)")
plt.show()
```

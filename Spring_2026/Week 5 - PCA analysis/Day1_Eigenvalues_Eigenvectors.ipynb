## 🔢 Step 3 — Covariance Matrix (2-Pixel Version)

**Theory:** The covariance matrix C captures how every pair of features (pixels) varies together.  
For n features it is **n×n**. Formula with centered data:

```
C = (1 / m−1) · Xᵀ X
```

We start with just **2 pixels** to keep the matrix small enough to read.

## 🧭 Step 4 — Eigenvalues & Eigenvectors

**Theory:** The eigenvalue equation: **C·v = λ·v**

```
v = eigenvector — a direction that doesn't rotate when multiplied by C
λ = eigenvalue  — how much that direction is stretched
```

For the covariance matrix: **largest λ = direction of maximum variance**.

## 📊 Step 7 — Eigenvalue Spectrum

**Theory:** The eigenvalue spectrum shows how much variance each component captures.  
The **"elbow"** of the cumulative curve reveals how many components are sufficient.

🟢 **PROVIDED** — run this.

````python
variance_pct = eigenvalues_full / eigenvalues_full.sum() * 100
cumulative   = np.cumsum(variance_pct)
````

````python
# --- YOUR CODE (~18 lines) ---

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Left — bar chart (top 30)
axes[0].bar(range(1, ___), variance_pct[:___], color="steelblue", alpha=0.8)
axes[0].set_xlabel("Principal Component")
axes[0].set_ylabel("Variance Explained (%)")
axes[0].set_title("Variance per Component (Top 30)")
axes[0].grid(True, alpha=0.3)

# Right — cumulative line (top 100)
axes[1].plot(range(1, ___), cumulative[:___], color="darkorange", lw=2)

for thresh in [80, 90, 95]:
    n = np.argmax(cumulative >= ___) + 1
    axes[1].axhline(___, ls="--", alpha=0.7,
                    label=f"{thresh}% → {n} components")
    print(f"{thresh}% variance needs {n} / {X.shape[1]} components")

axes[1].set_xlabel("Number of Components")
axes[1].set_ylabel("Cumulative Variance (%)")
axes[1].set_title("Cumulative Variance Explained")
axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout(); plt.show()

# ── HINT ────────────────────────────────────────────────────────────────
# axes[0].bar(range(1, 31), variance_pct[:___], ...)  # how many bars?
# n = np.argmax(cumulative >= ___) + 1   # what goes in argmax?
# axes[1].axhline(___, ...)              # what y-value for the line?
# ────────────────────────────────────────────────────────────────────────
````
